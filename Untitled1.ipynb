## Derin Öğrenme Modeli: DeepLabV3-ResNet50 ile LITS Üzerine İnce Ayar

Bu çalışmada, medikal görüntü segmentasyonu ve tümör tespiti alanında yaygın şekilde kullanılan bir model olan **DeepLabV3-ResNet50** mimarisi tercih edilmiştir. Model, hem doğruluk hem de uygulama kolaylığı açısından literatürde kendini kanıtlamış olup, özellikle karaciğer tümörü gibi sınırlı ve küçük alanları tespit etmede etkili sonuçlar vermektedir.

Modelimizi, **LITS (Liver Tumor Segmentation)** veri kümesine özgü olarak, özel bir şekilde ince ayar (fine-tune) ettik. Bu bağlamda iki farklı model eğitilmiştir:

1. **Vanilla DeepLabV3-ResNet50** modeli: Yüksek doğruluk sağlamaktadır ve FPGA üzerinde kullanılmak üzere optimize edilmiştir.
2. **Mobile DeepLabV3** modeli: Daha küçük ve hafif olmasına rağmen doğruluk açısından yetersiz bulunduğu için bu çalışmada tercih edilmemiştir.

Aşağıdaki hücrede, büyük (ve doğruluğu yüksek) olan model FPGA’ya yüklenmek üzere çağrılmakta ve parametre sayısı ile bellek kullanımı (FP32 formatında) görüntülenmektedir.


```python
# Büyük modelin yüklenmesi (DeepLabV3-ResNet50)
import os
from train import get_model_large

save_path = "./build/f_large_model.pth"
assert os.path.exists(save_path), "Model dosyası bulunamadı."

# 3 sınıf (karaciğer, tümör, arka plan) için model yükleniyor
model = get_model_large(num_classes=3, weights_path=save_path)

# Parametre sayısı ve FP32 boyutu
num_params = sum(p.numel() for p in model.parameters())
print(f"Toplam Parametre: {num_params:,}")
print(f"Model Boyutu (FP32): {num_params * 4 / 1e6:.2f} MB")
```

## Vitis AI için Model Kuantizasyonu: Wego_Torch ile INT8 Dönüşümü

KRIA KV260 FPGA kartımızın sınırlı hesaplama gücü ve belleği nedeniyle, PyTorch modelimizi doğrudan FP32 formatında çalıştırmak mümkün değil. Bu nedenle, modelimizi **wego_torch** kütüphanesi ile **INT8** formatına kuantize ediyoruz.

Kuantizasyon, model parametrelerinin ve hesaplamalarının bit derinliğini düşürerek bellek kullanımını yaklaşık 4 kat azaltır. Bu sayede:

- Model FPGA’da daha az yer kaplar,
- Hesaplama daha hızlı ve enerji verimli olur,
- Gerçek zamanlı tıbbi görüntü segmentasyonu mümkün hale gelir.

Vanilla DeepLabV3-ResNet50 modelimiz **yüksek doğruluk sağlarken**, mobil versiyon doğruluk kaybı nedeniyle tercih edilmemiştir. INT8 kuantizasyon ile model boyutunu küçültüp, performansı korumayı hedefliyoruz.


```python
from wego_torch import quantization, Linear, dtypes
model_int8 = quantization.quantize_dynamic(
    model,  # the original model
    {Linear},  # layers to quantize
    dtype=dtypes.int8  # data type
)

num_params = sum(p.numel() for p in model.parameters())
print(f"Params: {num_params:,} | INT8: {num_params * 1 / 1e6:.2f} MB")
```

## FPGA Üzerinde Görüntü Tabanlı Çıkarım: NIfTI Verisiyle Deneme

Bu bölümde, örnek bir **NIfTI** dosyası üzerinden **KRIA FPGA donanımı** üzerinde çıkarım (inference) gerçekleştiriyoruz. Buradaki amaç, medikal görüntü tanı sistemlerinin FPGA üzerinde çalıştırılabilirliğini ve verimliliğini göstermektir.

> ⚠️ Not: Bu çıkarım deneysel amaçlıdır ve tıbbi geçerlilik taşımaz. Yalnızca sistemin performansını test etmek için yapılmıştır.

Veri kümesinde yer alan `volume-12.nii` ve `volume-14.nii` örnekleriyle çalışıyoruz. Her bir hacim için:

- En büyük tümör bölgesinin bulunduğu **aksiyel düzlem** dilimi otomatik olarak çıkarılır.
- Bu 2D görüntü, modelimize giriş olarak verilir.
- Segmentasyon sonuçları görsel olarak değerlendirilir.

Burada yapılan çıkarım yalnızca **tek bir 2D dilim** üzerinden gerçekleştirilmiştir. Bu, tam hacimli bir çıkarım değildir ancak gerçek zamanlı çalışabilirliğin kanıtı olarak yeterlidir.

Modelimiz, bu örneklerde yüksek doğrulukta segmentasyon üretmiş ve FPGA üzerinde hızlı inference ile etkileyici bir performans göstermiştir.


```python
import matplotlib.pyplot as plt
import numpy as np
from dataset import create_sample_from_nii
from wego_torch import softmax
import wego_torch
# 12, 14
num = 14
# Load and visualize both the image and mask NIfTI files
image_path = f"./dataset/nii/volume-{num}.nii"
mask_path = f"./dataset/nii/segmentation-{num}.nii"

sample = create_sample_from_nii(image_path, mask_path)
sample["mask"].shape, sample["image"].shape

plt.imshow(np.rot90(sample["image"][0]))
plt.show()
plt.imshow(np.rot90(sample["mask"][0]))
plt.title("ground truth")
plt.show()

# add batch dimension to all
image = sample["image"].unsqueeze(0)
image = image.repeat(4, 1, 1, 1)
mask = sample["mask"].unsqueeze(0)

print(image.shape)
with wego_torch.inference():
    out = model(image)
    print(out.keys())
    out = out["out"]
    out = softmax(out, dim=1)
    out = np.argmax(out.n(), axis=1)[0]
    plt.imshow(np.rot90(out))
    plt.title("predictions")
    plt.show()
    print(np.unique(out))
```
